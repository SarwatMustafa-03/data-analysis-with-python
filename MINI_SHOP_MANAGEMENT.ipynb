# 🛒 Python Assignment 2
# Mini Shop Management System
# This program manages products, stock and sales.

```python
products = {}

sales = []

categories = ("Stationery", "Electronics", "Grocery")

unique_categories = set()
```

```python
def add_product():
    product_id = input("Enter Product ID: ")
    name = input("Enter Product Name: ")
    price = float(input("Enter Price: "))
    quantity = int(input("Enter Quantity: "))

    products[product_id] = {
        "name": name,
        "price": price,
        "quantity": quantity
    }

    print("✅ Product added successfully!")
```

```python
def view_products():
    if not products:
        print("No products available.")
        return

    print("\n========== PRODUCTS ==========")

    for product_id, product in products.items():
        print("ID:", product_id)
        print("Name:", product["name"])
        print("Price: Rs.", product["price"])
        print("Stock:", product["quantity"])
        print("----------------------")
```

```python
def search_product():
    product_id = input("Enter Product ID: ")

    if product_id in products:
        product = products[product_id]

        print("\nProduct Found! ✅")
        print("Name:", product["name"])
        print("Price: Rs.", product["price"])
        print("Stock:", product["quantity"])
    else:
        print("❌ Product not found!")
```

```python
def sell_product():
    product_id = input("Enter Product ID: ")

    if product_id not in products:
        print("❌ Product not found!")
        return

    quantity = int(input("Enter Quantity: "))

    product = products[product_id]

    if quantity > product["quantity"]:
        print("❌ Insufficient stock!")
        return

    total = quantity * product["price"]

    product["quantity"] -= quantity

    sales.append({
        "product": product["name"],
        "quantity": quantity,
        "total": total
    })

    print("\n========== BILL ==========")
    print("Product:", product["name"])
    print("Quantity:", quantity)
    print("Price: Rs.", product["price"])
    print("Total: Rs.", total)
    print("✅ Sale completed!")
```

```python
def low_stock():
    low_stock_products = [
        product["name"]
        for product in products.values()
        if product["quantity"] < 5
    ]

    print("\n⚠️ LOW STOCK PRODUCTS")

    if low_stock_products:
        for product_name in low_stock_products:
            print(product_name)
    else:
        print("No low-stock products.")
```

```python
def sales_summary():
    total_products_sold = sum(
        sale["quantity"] for sale in sales
    )

    total_sales_amount = sum(
        sale["total"] for sale in sales
    )

    print("\n========== SALES SUMMARY ==========")
    print("Total Products Sold:", total_products_sold)
    print("Total Sales Amount: Rs.", total_sales_amount)
```

```python
def product_analysis():
    product_prices = {
        product_id: product["price"]
        for product_id, product in products.items()
    }

    print("\nProduct Prices:")
    print(product_prices)

    sorted_products = sorted(
        products.values(),
        key=lambda x: x["price"]
    )

    print("\nProducts Sorted by Price:")

    for product in sorted_products:
        print(product["name"], "-", product["price"])
```

```python
def count_items(items):
    if len(items) == 0:
        return 0

    return 1 + count_items(items[1:])
```

```python
def show_sales_details():
    for product_id, product in products.items():

        for sale in sales:
            if sale["product"] == product["name"]:
                print(
                    product["name"],
                    "->",
                    sale["quantity"],
                    "items sold"
                )
```

```python
while True:

    print("\n================================")
    print("       🛒 MINI SHOP SYSTEM")
    print("================================")
    print("1. Add Product")
    print("2. View Products")
    print("3. Search Product")
    print("4. Sell Product")
    print("5. Check Low Stock")
    print("6. Sales Summary")
    print("7. Product Analysis")
    print("8. Sales Details")
    print("9. Count Sales Items")
    print("10. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        view_products()

    elif choice == "3":
        search_product()

    elif choice == "4":
        sell_product()

    elif choice == "5":
        low_stock()

    elif choice == "6":
        sales_summary()

    elif choice == "7":
        product_analysis()

    elif choice == "8":
        show_sales_details()

    elif choice == "9":
        print("Total Sales Records:", count_items(sales))

    elif choice == "10":
        print("Thank you for using Mini Shop System! 👋")
        break

    else:
        print("❌ Invalid choice!")
        continue
```
