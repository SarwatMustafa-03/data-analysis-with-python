```python
sub1=int(input("enter maths marks:"))
sub2=int(input("enter phys marks:"))
sub3=int(input("enter chem marks:"))
percentage = (sub1+sub2+sub3)/300*100
print(f"Percentage: {percentage:.2f}%")
if percentage>=80:
  print("A+")
elif percentage>=70:
  print("A")
elif percentage>=60:
  print("B")
elif percentage>=50:
  print("C")
else:
  print("D")
```
