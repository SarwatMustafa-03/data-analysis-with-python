```python
# Online Shopping Discount Calculator

# 1. Ask for shopping amount
shopping_amount = float(input("Enter your total shopping amount (Rs.): "))

# 2. Ask if customer is a Premium Member
premium_member = input("Are you a Premium Member? (Yes/No): ")

# 3. Calculate regular discount using if-elif-else

if shopping_amount >= 10000:
    discount_rate = 0.20
elif shopping_amount >= 5000:
    discount_rate = 0.10
elif shopping_amount >= 2000:
    discount_rate = 0.05
else:
    discount_rate = 0

# Calculate regular discount amount
regular_discount = shopping_amount * discount_rate

# 4. Nested if to check Premium Member
premium_discount = 0

if premium_member.lower() == "yes":
    if shopping_amount > 0:
        premium_discount = shopping_amount * 0.05

# 5. Calculate total discount
total_discount = regular_discount + premium_discount

# 6. Calculate final amount
final_amount = shopping_amount - total_discount

# Display results
print("\n--- Shopping Discount Summary ---")
print("Original Shopping Amount: Rs.", shopping_amount)
print("Regular Discount: Rs.", regular_discount)
print("Premium Discount: Rs.", premium_discount)
print("Total Discount: Rs.", total_discount)
print("Final Amount to Pay: Rs.", final_amount)
```
