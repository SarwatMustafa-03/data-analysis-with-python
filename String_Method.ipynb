##TO REMOVE SPACE,COVERT CASE,REPLACE WORDS,SPECIFIC WORD IS EXITING AUR NOT

```python
transaction = "Payment received from ALI MART - Karachi"
# The original code produced an AttributeError because .split('-') returns a list,
# and string methods like .lower() cannot be called on a list.
# Also, the .lower() method does not accept any arguments.
# To perform the intended string manipulations, apply them directly to the string.
# The order of operations is important for the desired outcome.
# Assuming the goal is to: remove spaces, convert to lowercase, replace words, and then check a prefix.
processed_transaction = transaction.strip().lower().replace("ali mart", "abc store")
print(processed_transaction.startswith("payment"))
```
