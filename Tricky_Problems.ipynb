##most improtant problem

```python
sales = [12000, 8500, 15000, 6000, 9500, 18000, 7000]
print(sum(sales))                   # fortotal
print(sum(sales) / len(sales))      # foravgerage
print(max(sales))                   # maxsale
print(min(sales))                    # minsale


for i in sales:
  if i<=10000:
    print(i)



average = sum(sales) / len(sales) # Calculate average and use correct variable name

if average > 10000:
    print("Good Sales Performance")
else:
    print("Sales Need Improvement")
```
